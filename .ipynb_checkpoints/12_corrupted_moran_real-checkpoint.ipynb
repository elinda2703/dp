```
import pandas as pd
import geopandas as gpd
from pathlib import Path
from libpysal import graph
from esda.moran import Moran
from joblib import Parallel, delayed
import numpy as np
import re
import os
import glob
import pyarrow as pa
import pyarrow.parquet as pq
```

```
sac_variables = {
    "sacramento": 'HH_INC',
    "elections": "pct_dem_12",
    "guerry": "Donatns",
}

c_runs = 10
N_SAMPLE = 20
N_JOBS = -1

GRAPH_NAME_RE = re.compile(r"g_missing(\d+)_run(\d+)\.parquet$")
```

```
def get_available_intensities(name, method, c_run):
    # get number of missing edges/corruption intensities for specific dataset/method
    folder = Path(f"graphs/real/{name}/no_isolates/{method}")
    pattern = re.compile(rf"g_missing(\d+)_run{c_run}\.parquet$")
    return sorted(
        int(pattern.match(f.name).group(1))
        for f in folder.glob(f"g_missing*_run{c_run}.parquet")
        if pattern.match(f.name)
    )

def parse_missing_edges(graph_path):
    # Derive the missing-edge count from the graph filename
    m = GRAPH_NAME_RE.search(str(graph_path))
    if not m:
        raise ValueError(f"could not parse missing-edge count from: {graph_path}")
    return int(m.group(1))


def select_sample_graphs(available_intensities, n_sample=20):
    """Linearly sample intensities from what's actually available.
    Always keeps the most-corrupted (max) graph."""
    available_intensities = sorted(available_intensities)
    max_idx = len(available_intensities) - 1

    idxs = np.linspace(0, max_idx, n_sample, dtype=int)
    idxs = idxs[1:]  # drop the least-corrupted graph
    idxs = sorted(set(idxs))

    return [available_intensities[i] for i in idxs]
```

```
def _process_real(name, var_col, method, c_run, intensity, gdf):
    graph_path = (
        f"graphs/real/{name}/no_isolates/{method}/"
        f"g_missing{intensity}_run{c_run}.parquet"
    )
    missing_edges = parse_missing_edges(graph_path) # get number of missing edges

    g = graph.read_parquet(graph_path)
    wm = g.to_W()
    wm.transform = 'r'
    id_order = list(wm.id_order)
    gdf_local = gdf.iloc[id_order]

    y = gdf_local[var_col].values
    mi = Moran(y, wm)

    return {
        "dataset": name, "variable": var_col, "corruption_method": method,
        "missing_edges": missing_edges, "corruption_run": c_run,
        "moran_i": mi.I, "moran_p": mi.p_sim, "moran_z": mi.z_sim,
    }
```

```
METHOD = 'random'  # all three real datasets get random removal

for name, var_col in sac_variables.items():
    gdf = gpd.read_parquet(f"data/real/{name}.parquet").reset_index(drop=True)

    jobs = [
        (c_run, intensity)
        for c_run in range(c_runs)
        for intensity in select_sample_graphs(
            get_available_intensities(name, METHOD, c_run), n_sample=N_SAMPLE
        )
    ]

    global_dir = Path(f"results/autocorrelation/real/{METHOD}/global")
    results = Parallel(n_jobs=N_JOBS, verbose=10)(
        delayed(_process_real)(name, var_col, METHOD, c_run, intensity, gdf)
        for c_run, intensity in jobs
    )

    global_df = pd.DataFrame(results)
    global_df.to_parquet(global_dir / f"{name}.parquet")
```

```
def _parse_filename(method, fname):
    """
    local (new, multi-run):   g_missing{i}_region_{region}_run{n}.parquet
    local (old, single-run):  g_missing{i}_run{region}.parquet
    borders / random:         g_missing{i}_run{n}.parquet
    """
    if method == "local":
        m = re.match(r"g_missing(\d+)_region_(.+)_run(\d+)\.parquet$", fname)
        if m:
            return int(m.group(1)), m.group(2), m.group(3)
        m = re.match(r"g_missing(\d+)_run(.+)\.parquet$", fname)
        if m:
            return int(m.group(1)), m.group(2), None
        return None, None, None
    else:
        m = re.match(r"g_missing(\d+)_run(.+)\.parquet$", fname)
        if m:
            return int(m.group(1)), None, m.group(2)
        return None, None, None


def _process_real_global(name, var_col, method, condition, region, run, intensity, graph_path, gdf):
    g = graph.read_parquet(graph_path)
    wm = g.to_W()
    wm.transform = 'r'
    id_order = list(wm.id_order)
    gdf_local = gdf.iloc[id_order]

    y = gdf_local[var_col].values
    mi = Moran(y, wm)

    return {
        "dataset": name, "variable": var_col, "corruption_method": method,
        "condition": condition, "region": region, "run": run,
        "missing_edges": intensity,
        "moran_i": mi.I, "moran_p": mi.p_sim, "moran_z": mi.z_sim,
    }


def process_guerry_all_global(
    var_col,
    gdf,
    conditions=("no_isolates",),
    methods=("borders", "local", "random"),
    out_dir="results/autocorrelation/real",
):
    """Runs global Moran's I on every corrupted graph saved for guerry -
    no sampling, across all conditions and all corruption methods."""
    name = "guerry"
    os.makedirs(out_dir, exist_ok=True)
    rows = []

    for condition in conditions:
        for method in methods:
            method_dir = f"graphs/real/{name}/{condition}/{method}"
            if not os.path.isdir(method_dir):
                continue

            paths = sorted(glob.glob(f"{method_dir}/g_missing*.parquet"))
            for graph_path in paths:
                fname = os.path.basename(graph_path)
                intensity, region, run = _parse_filename(method, fname)
                if intensity is None:
                    continue
                row = _process_real_global(
                    name=name, var_col=var_col, method=method, condition=condition,
                    region=region, run=run, intensity=intensity, graph_path=graph_path,
                    gdf=gdf,
                )
                rows.append(row)

    table = pa.Table.from_pylist(rows)
    out_path = f"{out_dir}/{name}_extra.parquet"
    pq.write_table(table, out_path)
    return rows, out_path
```

```
gdf = gpd.read_parquet("data/real/guerry.parquet").reset_index(drop=True)
rows, out_path = process_guerry_all_global(var_col="Donatns", gdf=gdf)
```
