```python
import itertools
import pandas as pd
import geopandas as gpd
from pathlib import Path
from libpysal import graph
from esda.moran import Moran, Moran_Local
from esda.geary import Geary
from esda.geary_local import Geary_Local
from joblib import Parallel, delayed
import numpy as np
```

```python
shapes = ['square','pent','hex']
sizes = [25,100,400]
n_runs = 10
rhos = np.arange(-0.9, 1.0, 0.1)
```

```python
def compute_truth_autocorrelation_synthetic(shape, size, rho, n_run, gdf):
    """
    Computes global + local Moran's I and Geary's C for the TRUE
    (uncorrupted) graph of a synthetic dataset, for one (rho, n_run) combo.
    Returns the global result dict and per-unit local arrays aligned to
    gdf's full row order (NaN for isolates).
    """
    graph_path = f"graphs/{shape}/size_{size}/g_true.parquet"
    g = graph.read_parquet(graph_path)
    wm = g.to_W()
    wm.transform = 'r'

    node_positions = list(wm.id_order)
    col = f"rho_{rho:.1f}_run_{n_run}"
    y = gdf.iloc[node_positions][col].values

    mi = Moran(y, wm)
    gc = Geary(y, wm)
    global_result = {
        "shape": shape, "size": size,
        "rho": rho, "data_run": n_run,
        "moran_i": mi.I, "moran_p": mi.p_sim, "moran_z": mi.z_sim,
    }

    return global_result
```

```python
for shape, size in itertools.product(shapes, sizes):
    src_path = f"data/autocorrelation/gdf_{shape}_{size}.parquet"
    gdf = gpd.read_parquet(src_path).reset_index(drop=True)

    truth_global_results = []

    for rho in rhos:
        for n_run in range(n_runs):
            global_result, local_arrays = compute_truth_autocorrelation_synthetic(
                shape, size, rho, n_run, gdf
            )
            truth_global_results.append(global_result)

    # save
    global_dir = Path(f"results/autocorrelation/synthetic/truth/global")
    global_dir.mkdir(parents=True, exist_ok=True)
    pd.DataFrame(truth_global_results).to_parquet(
        global_dir / f"{shape}_{size}.parquet"
    )
    print(f"Done: {shape} size_{size}")
```

```python
sac_variables ={"sacramento":'HH_INC',
                "elections":"pct_dem_12",
                "guerry":"Donatns"}
```

```python
def compute_truth_autocorrelation_real(name, var_col, gdf):
    graph_path = f"graphs/real/{name}/g_true.parquet"
    g = graph.read_parquet(graph_path)
    wm = g.to_W()
    wm.transform = 'r'

    node_ids = list(wm.id_order)
    y = gdf.loc[node_ids][var_col].values   # .loc not .iloc

    mi = Moran(y, wm)
    gc = Geary(y, wm)
    global_result = {
        "dataset": name,
        "variable": var_col,
        "moran_i": mi.I, "moran_p": mi.p_sim, "moran_z": mi.z_sim,
        "geary_c": gc.C, "geary_p": gc.p_sim, "geary_z": gc.z_sim,
    }

    mi_loc = Moran_Local(y, wm)
    gc_loc = Geary_Local(connectivity=wm).fit(y)

    n = len(gdf)
    local_moran_i = np.full(n, np.nan)
    local_moran_p = np.full(n, np.nan)
    local_moran_q = np.full(n, np.nan)
    local_geary_c = np.full(n, np.nan)
    local_geary_p = np.full(n, np.nan)

    # use positional location of node_ids within gdf's index
    positions = [gdf.index.get_loc(nid) for nid in node_ids]
    local_moran_i[positions] = mi_loc.Is
    local_moran_p[positions] = mi_loc.p_sim
    local_moran_q[positions] = mi_loc.q
    local_geary_c[positions] = gc_loc.localG
    local_geary_p[positions] = gc_loc.p_sim

    local_arrays = {
        "true_local_moran_i": local_moran_i,
        "true_local_moran_p": local_moran_p,
        "true_local_moran_q": local_moran_q,
        "true_local_geary_c": local_geary_c,
        "true_local_geary_p": local_geary_p,
    }
    return global_result, local_arrays

```

```python
truth_global_results = []
global_dir = Path("results/autocorrelation/real/truth/global")
global_dir.mkdir(parents=True, exist_ok=True)

for name, var_col in sac_variables.items():
    src_path = f"data/real/{name}.parquet"
    gdf = gpd.read_parquet(src_path)

    global_result, local_arrays = compute_truth_autocorrelation_real(name, var_col, gdf)
    truth_global_results.append(global_result)

    for col_name, values in local_arrays.items():
        gdf[col_name] = values

    gdf.to_parquet(src_path)
    print(f"Done: {name}")

pd.DataFrame(truth_global_results).to_parquet(global_dir / "global.parquet")
```
