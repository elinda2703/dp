```python
base = Path("results/autocorrelation/synthetic")
corruption_methods = ["random", "local", "borders"]
dfs = []
for method in corruption_methods:
    folder = base / method / "global"
    for file in sorted(folder.glob("*.parquet")):
        if file.stem == "all_extra":
            continue
        dfs.append(pd.read_parquet(file))
```

```python
cor = pd.concat(dfs, ignore_index=True)
```

```python
truth_folder = Path("results/autocorrelation/synthetic/truth/global")

dfs = [pd.read_parquet(file) for file in sorted(truth_folder.glob("*.parquet"))]
truth = pd.concat(dfs, ignore_index=True)
```

```python
truth_small = truth[['shape', 'size', 'rho', 'data_run', 'moran_i', 'moran_p']].rename(
    columns={'moran_i': 'true_moran_i', 'moran_p': 'true_moran_p'}
)

cor = cor.merge(truth_small, on=['shape', 'size', 'rho', 'data_run'], how='left')
cor['moran_i_deviation'] = cor['moran_i'] - cor['true_moran_i']

# sanity check: every row should have found a match
n_unmatched = cor['true_moran_i'].isna().sum()
print(f"{n_unmatched} rows with no matching ground truth (out of {len(cor)})")

cor['moran_i_abs_deviation'] = (cor['moran_i'] - cor['true_moran_i']).abs()
```

```python
def true_edge_count(shape, size):
    path = f"graphs/{shape}/size_{size}/g_true.parquet"
    g_true = graph.read_parquet(path)
    return g_true.n_edges / 2  # stored bidirectionally, but graph is undirected

# only load each true graph once, not once per row
unique_pairs = cor[['shape', 'size']].drop_duplicates()
edge_df = pd.DataFrame(
    [(s, sz, true_edge_count(s, sz)) for s, sz in unique_pairs.itertuples(index=False)],
    columns=['shape', 'size', 'n_true_edges']
)

cor = cor.merge(edge_df, on=['shape', 'size'], how='left')
cor['percent_edges_missing'] = cor['missing_edges'] / cor['n_true_edges'] * 100

# sanity check: every row should have found a true-edge count
n_unmatched = cor['n_true_edges'].isna().sum()
print(f"{n_unmatched} rows with no matching true graph (out of {len(cor)})")
print(edge_df)
```

```python
subset = cor.copy()

# --- bin into 5% intervals ---
bin_width = 5
max_pct = subset['percent_edges_missing'].max()
bin_edges = np.arange(0, np.ceil(max_pct / bin_width) * bin_width + bin_width, bin_width)
subset['missing_bin'] = pd.cut(subset['percent_edges_missing'], bins=bin_edges,
                                right=False, include_lowest=True)

def fmt_bin(iv):
    return f"{iv.left:g}-{iv.right:g}"

subset['missing_bin_label'] = subset['missing_bin'].map(fmt_bin)
bin_order = [fmt_bin(iv) for iv in subset['missing_bin'].cat.categories]

# --- cut off bins beyond 35-40 ---
cutoff_label = "30-35"
cutoff_idx = bin_order.index(cutoff_label)
bin_order = bin_order[:cutoff_idx + 1]
subset = subset[subset['missing_bin_label'].isin(bin_order)]

# --- aggregate: mean and 95th percentile per shape/size/method/bin ---
grouped = (
    subset.groupby(['shape', 'size', 'corruption_method', 'missing_bin_label'], observed=True)
    ['moran_i_abs_deviation']
    .agg(mean='mean', p95=lambda x: np.percentile(x, 95))
    .reset_index()
)
grouped['missing_bin_label'] = pd.Categorical(
    grouped['missing_bin_label'], categories=bin_order, ordered=True
)
grouped = grouped.sort_values(['shape', 'size', 'corruption_method', 'missing_bin_label'])

shapes = ['hex', 'pent', 'square']
sizes = [100, 400]
methods = subset['corruption_method'].unique()
method_colors = dict(zip(methods, plt.cm.tab10.colors))

fig, axes = plt.subplots(3, 2, figsize=(8.27, 11.0), sharex=True, sharey=True)

for i, shape in enumerate(shapes):
    for j, size in enumerate(sizes):
        ax = axes[i, j]
        d = grouped[(grouped['shape'] == shape) & (grouped['size'] == size)]

        for method in methods:
            m = d[d['corruption_method'] == method]
            color = method_colors[method]
            ax.plot(m['missing_bin_label'], m['mean'], marker='o', markersize=3.5,
                    linewidth=1.3, linestyle='-', color=color, label=f'{method} (mean)')
            ax.plot(m['missing_bin_label'], m['p95'], marker='s', markersize=3.5,
                    linewidth=1.3, linestyle='--', color=color, label=f'{method} (p95)')

        ax.set_title(f"{shape}_{size}", fontsize=10)
        ax.tick_params(axis='x', rotation=90, labelsize=6)
        ax.tick_params(axis='y', labelsize=8)

        # --- finer gridlines ---
        ax.yaxis.set_minor_locator(AutoMinorLocator(4))
        ax.grid(which='major', axis='y', alpha=0.35, linewidth=0.6)
        ax.grid(which='minor', axis='y', alpha=0.15, linewidth=0.4, linestyle=':')
        ax.grid(which='major', axis='x', alpha=0.2, linewidth=0.6)

        row, col = i, j
        ax.set_xlabel('% edges missing' if row == 2 else '', fontsize=9)
        ax.set_ylabel("|Moran's I deviation|" if col == 0 else '', fontsize=9)
        if col != 0:
            ax.tick_params(axis='y', labelleft=False)

        if i == 0 and j == 0:
            ax.legend(fontsize=6, loc='upper left', ncol=1)
sns.despine()
#fig.suptitle("Absolute Moran's I deviation by spatial layout and corruption method", fontsize=13, y=0.995)
fig.tight_layout(rect=[0.02, 0.01, 0.98, 0.965])
plt.savefig("figs/moran_deviation_methods.png", dpi=600, bbox_inches="tight")
```

```python
subset = cor.copy()

# --- bin into 5% intervals ---
bin_width = 5
max_pct = subset['percent_edges_missing'].max()
bin_edges = np.arange(0, np.ceil(max_pct / bin_width) * bin_width + bin_width, bin_width)
subset['missing_bin'] = pd.cut(subset['percent_edges_missing'], bins=bin_edges,
                                right=False, include_lowest=True)

def fmt_bin(iv):
    return f"{iv.left:g}-{iv.right:g}"

subset['missing_bin_label'] = subset['missing_bin'].map(fmt_bin)
bin_order = [fmt_bin(iv) for iv in subset['missing_bin'].cat.categories]

# --- cut off bins beyond 35-40 ---
cutoff_label = "35-40"
cutoff_idx = bin_order.index(cutoff_label)
bin_order = bin_order[:cutoff_idx + 1]
subset = subset[subset['missing_bin_label'].isin(bin_order)]

# --- aggregate: mean and max per shape/size/method/bin ---
grouped = (
    subset.groupby(['shape', 'size', 'corruption_method', 'missing_bin_label'], observed=True)
    ['moran_i_abs_deviation']
    .agg(mean='mean', max='max')
    .reset_index()
)
grouped['missing_bin_label'] = pd.Categorical(
    grouped['missing_bin_label'], categories=bin_order, ordered=True
)
grouped = grouped.sort_values(['shape', 'size', 'corruption_method', 'missing_bin_label'])

shapes = ['hex', 'pent', 'square']
sizes = [100, 400]
methods = subset['corruption_method'].unique()
method_colors = dict(zip(methods, plt.cm.tab10.colors))

fig, axes = plt.subplots(3, 2, figsize=(8.27, 11.0), sharex=True, sharey=True)

for i, shape in enumerate(shapes):
    for j, size in enumerate(sizes):
        ax = axes[i, j]
        d = grouped[(grouped['shape'] == shape) & (grouped['size'] == size)]

        for method in methods:
            m = d[d['corruption_method'] == method]
            color = method_colors[method]
            ax.plot(m['missing_bin_label'], m['mean'], marker='o', markersize=3.5,
                    linewidth=1.3, linestyle='-', color=color, label=f'{method} (mean)')
            ax.plot(m['missing_bin_label'], m['max'], marker='s', markersize=3.5,
                    linewidth=1.3, linestyle='--', color=color, label=f'{method} (max)')

        ax.set_title(f"{shape}_{size}", fontsize=10)
        ax.tick_params(axis='x', rotation=90, labelsize=6)
        ax.tick_params(axis='y', labelsize=8)

        # --- finer gridlines ---
        ax.yaxis.set_minor_locator(AutoMinorLocator(4))
        ax.grid(which='major', axis='y', alpha=0.35, linewidth=0.6)
        ax.grid(which='minor', axis='y', alpha=0.15, linewidth=0.4, linestyle=':')
        ax.grid(which='major', axis='x', alpha=0.2, linewidth=0.6)

        row, col = i, j
        ax.set_xlabel('% edges missing' if row == 2 else '', fontsize=9)
        ax.set_ylabel("|Moran's I deviation|" if col == 0 else '', fontsize=9)
        if col != 0:
            ax.tick_params(axis='y', labelleft=False)

        if i == 0 and j == 0:
            ax.legend(fontsize=6, loc='upper left', ncol=1)

fig.suptitle("Absolute Moran's I deviation by spatial layout and corruption method", fontsize=13, y=0.995)
fig.tight_layout(rect=[0.02, 0.01, 0.98, 0.965])
#plt.savefig("figs/moran_deviation.png", dpi=600, bbox_inches="tight")
```

```python
df = cor[cor['corruption_method'] == 'random'].copy()
 
# --- bin by the *original* (ground truth) Moran's I ---
bins = [-np.inf, -0.1, 0.1, np.inf]
bin_labels = ['< -0.1', '-0.1 to 0.1', '> 0.1']
bin_colors = {'< -0.1': 'tab:blue', '-0.1 to 0.1': 'gray', '> 0.1': 'tab:red'}
df['true_moran_bin'] = pd.cut(df['true_moran_i'], bins=bins, labels=bin_labels)
 
shapes = sorted(df['shape'].unique())
sizes = sorted(df['size'].unique())
print(f"shapes found: {shapes}")
print(f"sizes found: {sizes}")
print(f"-> {len(shapes)} x {len(sizes)} = {len(shapes) * len(sizes)} subplots")
 
fig, axes = plt.subplots(
    len(shapes), len(sizes),
    figsize=(5 * len(sizes), 4 * len(shapes)),
    sharex=True, sharey=True
)
axes = np.atleast_2d(axes)
 
for i, shape in enumerate(shapes):
    for j, size in enumerate(sizes):
        ax = axes[i, j]
        sub = df[(df['shape'] == shape) & (df['size'] == size)]
 
        if sub.empty:
            ax.set_visible(False)
            continue
 
        for b in bin_labels:
            data = sub.loc[sub['true_moran_bin'] == b, 'moran_i_deviation'].dropna()
            if len(data) > 1:
                sns.kdeplot(data, label=b, ax=ax, color=bin_colors[b], clip=(-0.3,0.3))
 
        ax.axvline(0, color='gray', linestyle='--', linewidth=0.8)
        ax.set_title(f"{shape} - size {size}")
        ax.set_xlabel("deviation")
        if j == 0:
            ax.set_ylabel("density")
        else:
            ax.set_ylabel("")
        ax.legend(title="true Moran's I", fontsize='small')

plt.tight_layout()
plt.savefig("figs/moran_deviation_kde.png", dpi=600, bbox_inches='tight')
```

```python
# n_true_edges: purely structural -> shape + size
edge_lookup = cor_extra[["shape", "size", "n_true_edges"]].drop_duplicates()
assert not edge_lookup.duplicated(subset=["shape", "size"]).any(), \
    "n_true_edges varies within a shape/size — assumption wrong"

# true_moran_i: depends on the data draw, not on corruption -> shape + size + rho + data_run
moran_lookup = cor_extra[["shape", "size", "rho", "data_run", "true_moran_i"]].drop_duplicates()
assert not moran_lookup.duplicated(subset=["shape", "size", "rho", "data_run"]).any(), \
    "true_moran_i varies within a data draw — assumption wrong"

# merge both into extra
extra = extra.merge(edge_lookup, on=["shape", "size"], how="left")
extra = extra.merge(moran_lookup, on=["shape", "size", "rho", "data_run"], how="left")

# derive the remaining columns
extra["percent_edges_missing"] = extra["missing_edges"] / extra["n_true_edges"] * 100
extra["moran_i_deviation"] = extra["moran_i"] - extra["true_moran_i"]
extra["moran_i_abs_deviation"] = extra["moran_i_deviation"].abs()

# sanity check nothing failed to match
assert extra["true_moran_i"].notna().all() and extra["n_true_edges"].notna().all()

# concat with the full dataset
cor_full = pd.concat([cor_extra, extra], ignore_index=True)
```

```python
SHAPES = ["square", "pent", "hex"]
BIN_WIDTH = 0.2

# --- 1. prepare data ---
df_100 = cor.loc[(cor["size"] == 100) & (cor["corruption_method"] == "random")].copy()
df_100["abs_deviation"] = df_100["moran_i_deviation"].abs()

# centered bins: shift edges by half a bin width, then clip to [-1, 1]
half = BIN_WIDTH / 2
bin_edges = np.arange(-1.0 - half, 1.0 + half + BIN_WIDTH, BIN_WIDTH)
bin_edges = np.clip(bin_edges, -1.0, 1.0)
bin_edges = np.unique(np.round(bin_edges, 6))  # round first to avoid float-noise duplicates

df_100["moran_bin"] = pd.cut(df_100["true_moran_i"], bins=bin_edges)

grouped = (
    df_100.groupby(["shape", "moran_bin"], observed=True)
    .agg(max_abs_dev=("abs_deviation", "max"))
    .reset_index()
)

# --- 2. pivot, then force correct column order explicitly ---
pivot = grouped.pivot(index="shape", columns="moran_bin", values="max_abs_dev")
pivot = pivot.reindex(SHAPES)

# rebuild the FULL sorted interval sequence from bin_edges, independent of what pivot did
full_intervals = pd.IntervalIndex.from_breaks(bin_edges)
pivot = pivot.reindex(columns=full_intervals)

# drop columns that are entirely empty across all shapes (e.g. the extreme -1.0 to -0.9 bin)
pivot = pivot.dropna(axis=1, how="all")

# space-padded sign so "-0.10" and " 0.10" line up under a monospace font
pivot.columns = [f"{iv.left: .2f} - {iv.right: .2f}" for iv in pivot.columns]

# --- 3. grey-out NaN helper ---
def grey_nan(val):
    return "background-color: #d9d9d9; color: #d9d9d9;" if pd.isna(val) else ""

# --- 4. style it ---
styled = (
    pivot.style
    .background_gradient(cmap="YlOrRd", axis=None)
    .format("{:.3f}", na_rep="—")
    .set_properties(**{"text-align": "center", "border": "1px solid #ddd", "padding": "6px 10px"})
    .set_table_styles([
        {"selector": "th.col_heading", "props": [
            ("text-align", "center"),
            ("background-color", "#f5f5f5"),
            ("padding", "2px 10px 2px 10px"),
            ("vertical-align", "bottom"),
            ("font-family", "monospace"),
            ("white-space", "pre"),
        ]},
        {"selector": "th.row_heading", "props": [
            ("text-align", "left"), ("padding", "6px 10px"),
        ]},
        {"selector": "th.index_name", "props": [
            ("padding", "0px 10px 6px 10px"),
            ("vertical-align", "top"),
            ("text-align", "left"),
        ]},
    ])
)

styled = styled.map(grey_nan) if hasattr(styled, "map") else styled.applymap(grey_nan)
styled
```

```python
TRUE_PVAL_COL = 'true_moran_p'
PVAL_COL = 'moran_p'
BIN_WIDTH_PCT = 5
# --- 1. prepare data: random method only ---
subset_random = cor[cor['corruption_method'] == 'random'].copy()
subset_random['is_true_sig'] = subset_random[TRUE_PVAL_COL] < 0.05
subset_random['is_calc_sig'] = subset_random[PVAL_COL] < 0.05
pct_bin_edges = np.arange(0, 100 + BIN_WIDTH_PCT, BIN_WIDTH_PCT)
subset_random['pct_bin'] = pd.cut(subset_random['percent_edges_missing'], bins=pct_bin_edges)
label_nonsig = "significance gain"
label_falsenonsig = "significance loss"
shapes = sorted(subset_random['shape'].unique())
sizes = sorted(subset_random['size'].unique())
combos = [(s, sz) for s in shapes for sz in sizes]
print(f"{len(combos)} shape/size combos -> {len(combos)} lines per subplot")

# color = shape (fixed palette from earlier), linestyle = size
SHAPE_COLORS = {"square": "#d95f02", "pent": "#1b9e77", "hex": "#08519c"}  # orange, green, blue
size_linestyles = {sz: ls for sz, ls in zip(sizes, ['-', '--', ':', '-.'])}

# --- 2. plot: one subplot for false-sig, one for false-non-sig, all combos overlaid ---
fig, (ax_fs, ax_fns) = plt.subplots(1, 2, figsize=(11,6), sharex=True, sharey=True)
for shape_val, size_val in combos:
    g_df_all = subset_random[(subset_random['shape'] == shape_val) & (subset_random['size'] == size_val)]
    if g_df_all.empty:
        continue
    color = SHAPE_COLORS[shape_val]
    linestyle = size_linestyles[size_val]
    line_label = f"{shape_val}, n={size_val}"
    # --- false significant: originally non-sig, now flagged sig ---
    nonsig_all = g_df_all[~g_df_all['is_true_sig']]
    results_nonsig = []
    for pct_bin, bin_df in nonsig_all.groupby('pct_bin', observed=True):
        if len(bin_df) > 0:
            mean_missing = bin_df['percent_edges_missing'].mean()
            pct_faked = (bin_df['is_calc_sig'] == True).sum() / len(bin_df) * 100
            results_nonsig.append({'mean_missing': mean_missing, 'Error Rate': pct_faked})
    df_nonsig = pd.DataFrame(results_nonsig).sort_values('mean_missing')
    # --- false non-significant: originally sig, now flagged non-sig ---
    sig_all = g_df_all[g_df_all['is_true_sig']]
    results_sig = []
    for pct_bin, bin_df in sig_all.groupby('pct_bin', observed=True):
        if len(bin_df) > 0:
            mean_missing = bin_df['percent_edges_missing'].mean()
            pct_lost = (bin_df['is_calc_sig'] == False).sum() / len(bin_df) * 100
            results_sig.append({'mean_missing': mean_missing, 'Error Rate': pct_lost})
    df_sig = pd.DataFrame(results_sig).sort_values('mean_missing')
    # anchor both lines at (0, 0) — no missing edges, no misclassification
    if not df_nonsig.empty:
        x = np.concatenate([[0], df_nonsig['mean_missing'].values])
        y = np.concatenate([[0], df_nonsig['Error Rate'].values])
        ax_fs.plot(x, y, color=color, linestyle=linestyle, linewidth=1.5, alpha=0.9, label=line_label)
    if not df_sig.empty:
        x = np.concatenate([[0], df_sig['mean_missing'].values])
        y = np.concatenate([[0], df_sig['Error Rate'].values])
        ax_fns.plot(x, y, color=color, linestyle=linestyle, linewidth=1.5, alpha=0.9, label=line_label)
for ax, title in [(ax_fs, label_nonsig), (ax_fns, label_falsenonsig)]:
    ax.set_title(title)
    ax.set_xlabel("edges missing (%)")
    ax.set_ylim(bottom=0)
    ax.grid(True, linestyle=':', alpha=0.6)
    sns.despine(ax=ax)
ax_fs.set_ylabel("significance misclassification (%)")
# --- legend on first subplot, 3x3 grid ---
ax_fs.legend(loc='upper left', bbox_to_anchor=(0.02, 1), frameon=False, fontsize=8, ncol=3,
             columnspacing=1.0, handletextpad=0.5)
plt.tight_layout()
Path("figs").mkdir(exist_ok=True)
plt.savefig("figs/p_values_two_panels_by_error_type.png", bbox_inches="tight", dpi=600)
#plt.show()
```

```python
BIN_WIDTH = 0.025  # histogram bin width for corrupted moran_i
MIN_DENOM = 5       # suppress bins where the denominator group has too few points to be meaningful
 
# --- 1. prepare data: random method only ---
df_all = cor.loc[cor["corruption_method"] == "random"].copy()
df_all["is_true_sig"] = df_all["true_moran_p"] < 0.05
df_all["is_calc_sig"] = df_all["moran_p"] < 0.05
df_all["false_sig"] = (~df_all["is_true_sig"]) & df_all["is_calc_sig"]      # truly non-sig, flagged sig
df_all["false_nonsig"] = df_all["is_true_sig"] & (~df_all["is_calc_sig"])  # truly sig, flagged non-sig
 
sizes = sorted(df_all["size"].unique())
shapes = sorted(df_all["shape"].unique())
print(f"sizes: {sizes} ({len(sizes)}) x shapes: {shapes} ({len(shapes)}) -> {len(sizes) * len(shapes)} subplots")
 
label_falsenonsig = "significance loss"
label_falsesig = "significance gain"
error_colors = {label_falsesig: "#4c72b0", label_falsenonsig: "#a50f15"}  # softer, muted blue / red
 
bin_edges = np.arange(-0.75, 0.75 + BIN_WIDTH, BIN_WIDTH)
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
bar_width = BIN_WIDTH * 0.9
 
 
def compute_bin_fractions(data, bin_edges, min_denom=MIN_DENOM):
    """For each moran_i bin, compute:
       - frac_false_nonsig = (# true-sig points flagged non-sig) / (# true-sig points) in that bin
       - frac_false_sig    = (# true-nonsig points flagged sig)  / (# true-nonsig points) in that bin
       Bins with denom < min_denom are set to NaN (too noisy to trust).
    """
    n_bins = len(bin_edges) - 1
    bin_idx = np.digitize(data["true_moran_i"], bin_edges) - 1
    frac_false_sig = np.full(n_bins, np.nan)
    frac_false_nonsig = np.full(n_bins, np.nan)
    n_true_sig = np.zeros(n_bins, dtype=int)
    n_true_nonsig = np.zeros(n_bins, dtype=int)
    for b in range(n_bins):
        bin_data = data[bin_idx == b]
        ts = bin_data[bin_data["is_true_sig"]]
        tns = bin_data[~bin_data["is_true_sig"]]
        n_true_sig[b] = len(ts)
        n_true_nonsig[b] = len(tns)
        if len(ts) >= min_denom:
            frac_false_nonsig[b] = ts["false_nonsig"].sum() / len(ts)
        if len(tns) >= min_denom:
            frac_false_sig[b] = tns["false_sig"].sum() / len(tns)
    return frac_false_sig, frac_false_nonsig, n_true_sig, n_true_nonsig
 
 
# --- 2. plot: grid of shape (rows) x size (columns), sized for A4 portrait ---
n_rows, n_cols = len(shapes), len(sizes)
 
# A4 = 8.27 x 11.69 in; leave a little margin for the legend/suptitle
fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(11, 11),
    sharex=True, sharey=True,
)
axes = np.atleast_2d(axes)
 
for i, shp in enumerate(shapes):
    for j, sz in enumerate(sizes):
        ax = axes[i, j]
        cell_data = df_all[(df_all["shape"] == shp) & (df_all["size"] == sz)]
        if cell_data.empty:
            ax.set_visible(False)
            continue
 
        frac_fs, frac_fns, n_ts, n_tns = compute_bin_fractions(cell_data, bin_edges)
 
        ax.bar(bin_centers, frac_fns * 100, width=bar_width,
               color=error_colors[label_falsenonsig], alpha=0.5,
               edgecolor=error_colors[label_falsenonsig], linewidth=0.5,
               zorder=2, label=label_falsenonsig)
        ax.bar(bin_centers, frac_fs * 100, width=bar_width,
               color=error_colors[label_falsesig], alpha=0.5,
               edgecolor=error_colors[label_falsesig], linewidth=0.5,
               zorder=2, label=label_falsesig)
 
        ax.set_title(f"{shp}, n = {sz}", pad=8, fontsize=9.5)
        if i == n_rows - 1:
            ax.set_xlabel("true Moran's I", fontsize=9)
        if j == 0:
            ax.set_ylabel("misclassification rate (%)", fontsize=9)
 
        ax.set_xlim(-0.75, 0.75)
        ax.set_ylim(0, 100)
        ax.tick_params(labelsize=8)
        ax.grid(True, linestyle=":", alpha=0.7, color="lightgrey")
        sns.despine(ax=ax)
 
# --- 3. legend (first subplot only) ---
handles = [mpatches.Patch(color=c, alpha=0.6, label=lbl) for lbl, c in error_colors.items()]
axes[0, 0].legend(handles=handles, loc="upper left", bbox_to_anchor=(0.02, 1),
                   frameon=False, fontsize=8)
 
plt.tight_layout()
 
Path("figs").mkdir(exist_ok=True)
plt.savefig("figs/moran_falsesig_falsenonsig_rate_by_size_shape_9panel.png",
            dpi=600, bbox_inches="tight")
```

```python
PALETTE_STYLE = {"mean": 2.5, "p95": 1.5}  # kept for reference, not directly used
BIN_WIDTH = 5  # percent_edges_missing bin width, e.g. 0-5, 5-10, ...
# --- 1. prepare data ---
df_all = cor.loc[cor["corruption_method"] == "random"].copy()
df_all["abs_deviation"] = df_all["moran_i_deviation"].abs()
# bin percent_edges_missing into fixed-width 5% buckets
bin_edges = np.arange(0, 100 + BIN_WIDTH, BIN_WIDTH)
df_all["percent_missing_bin"] = pd.cut(df_all["percent_edges_missing"], bins=bin_edges)
grouped = (
    df_all.groupby(["shape", "size", "percent_missing_bin"], observed=True)
    .agg(
        mean_percent_missing=("percent_edges_missing", "mean"),  # actual mean within the bin, for x position
        mean_abs_dev=("abs_deviation", "mean"),
        p95_abs_dev=("abs_deviation", lambda x: x.quantile(0.95)),
    )
    .reset_index()
)
# fixed color palette, one color per shape
SHAPE_COLORS = {"square": "#d95f02", "pent": "#1b9e77", "hex": "#08519c"}  # orange, green, blue
# --- 2. plot: one subplot per size ---
sizes = sorted(df_all["size"].unique())
fig, axes = plt.subplots(1, len(sizes), figsize=(16, 7), sharex=True, sharey=True)
if len(sizes) == 1:
    axes = [axes]
for ax, sz in zip(axes, sizes):
    subset = grouped[grouped["size"] == sz]
    for shape_val, color in SHAPE_COLORS.items():
        shape_data = subset[subset["shape"] == shape_val].sort_values("mean_percent_missing")
        if shape_data.empty:
            continue
        # force line to start at (0, 0) — no missing edges, no deviation
        x = np.concatenate([[0], shape_data["mean_percent_missing"].values])
        mean_y = np.concatenate([[0], shape_data["mean_abs_dev"].values])
        p95_y = np.concatenate([[0], shape_data["p95_abs_dev"].values])
        ax.plot(x, mean_y, color=color, linewidth=2.5)               # mean
        ax.plot(x, p95_y, color=color, linewidth=1.5, linestyle="--")  # upper 95%
        ax.fill_between(x, mean_y, p95_y, color=color, alpha=0.15, linewidth=0)
        # Maximum Edge Loss (Vertical Dotted Line)
        max_missing = shape_data["mean_percent_missing"].max()
        ax.axvline(x=max_missing, color=color, linestyle=":", linewidth=1)
    ax.set_title(f"n = {sz}", pad=15)
    ax.set_xlabel("edges missing (%)")

    # finer grid: minor ticks + minor gridlines, on top of major ones
    ax.xaxis.set_minor_locator(mticker.AutoMinorLocator(4))
    ax.yaxis.set_minor_locator(mticker.AutoMinorLocator(10))  # finer on y
    ax.grid(True, which="major", linestyle=":", alpha=0.7, color="lightgrey")
    ax.grid(True, which="minor", linestyle=":", alpha=0.35, color="lightgrey")

    sns.despine(ax=ax)
for a in axes:
    a.set_ylim(bottom=0)
axes[0].set_ylabel("absolute deviation")
# --- 3. legends (first subplot only) ---
style_handles = [
    mlines.Line2D([], [], color="black", linestyle="-", linewidth=2.5, label="mean"),
    mlines.Line2D([], [], color="black", linestyle="--", linewidth=1.5, label="upper 95% interval"),
    mlines.Line2D([], [], color="black", linestyle=":", linewidth=1, label="maximum edge loss"),
]
shape_handles = [mlines.Line2D([0], [0], color=c, linewidth=4, label=s) for s, c in SHAPE_COLORS.items()]
style_legend = axes[0].legend(
    handles=style_handles, loc="upper left", bbox_to_anchor=(0.02, 0.77),  # lowered slightly to fit 3rd item
    frameon=False, alignment="left"
)
axes[0].add_artist(style_legend)
axes[0].legend(
    handles=shape_handles, loc="upper left", bbox_to_anchor=(0.02, 1),
    title="lattice shape", frameon=False, alignment="left"
)
plt.tight_layout()
Path("figs").mkdir(exist_ok=True)
plt.savefig("figs/size_comparison.png", dpi=600, bbox_inches="tight")
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

# 5. Group by BOTH the corruption level (g_run) AND the new Moran's I bin
grouped = subset_corrupted_all.groupby(['g_run', 'true_moran_interval'], observed=False)

# 6. Calculate Mean Missing, Mean Error, AND Max Error (95th percentile)
error_data = grouped.agg(
    mean_missing=('real_missing', lambda x: x.mean() * 100),
    mean_error=('abs_deviation', 'mean'), 
    max_error=('abs_deviation', lambda x: x.quantile(0.95))
).reset_index()

# Drop any empty bins
error_data = error_data.dropna(subset=['mean_error', 'max_error'])

# 7. Create the Plot
fig, ax = plt.subplots(figsize=(8, 7))

for i, label in enumerate(labels):
    bin_data = error_data[error_data['true_moran_interval'] == label]
    
    # Only plot if there are actually samples in this interval
    if not bin_data.empty:
        x = bin_data['mean_missing']
        y_mean = bin_data['mean_error']
        y_max = bin_data['max_error']
        
        # 1. Plot the MEAN (Solid Line)
        ax.plot(
            x, 
            y_mean, 
            color=custom_palette[label], 
            linestyle='-', 
            linewidth=2.5
        )
        
        # 2. Fill the area ONLY ABOVE the mean (up to the 95th percentile)
        ax.fill_between(
            x, 
            y_mean,   # The bottom boundary (Mean)
            y_max,    # The top boundary (95th Percentile)
            color=custom_palette[label], 
            alpha=0.1, # Makes the shading beautifully transparent
            linewidth=0 # Removes the harsh border around the shaded shape
        )
        
        # 3. Plot the UPPER BOUNDARY (Dashed Line)
        ax.plot(
            x,
            y_max,
            color=custom_palette[label],
            linestyle='--',
            linewidth=1.5,
            alpha=0.8
        )

# Formatting
ax.set_xlabel("edges missing (%)", fontsize=12)
ax.set_ylabel("absolute deviation", fontsize=12)
#ax.set_title("title TBD", fontsize=14)

# Start y-axis at 0 
ax.set_ylim(bottom=0)
ax.grid(True, linestyle=':', alpha=0.6)

# Despine
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# --- CUSTOM COMPOSITE LEGEND ---
# 1. First, create handles for your True Moran's I categories (the colors)
legend_handles = []

# 2. Add a spacer (an invisible line) to visually separate categories from line styles
legend_handles.append(Line2D([0], [0], color='none', label=' ')) 

# 3. Add the black reference lines for the Mean and the 95% Interval
legend_handles.extend([
    Line2D([0], [0], color='black', linestyle='-', lw=2.5, label='mean'),
    Line2D([0], [0], color='black', linestyle='--', lw=1.5, alpha=0.8, label='upper 95% interval')
])

# Draw the legend onto the plot
ax.legend(handles=legend_handles, frameon=False,fontsize=10)

plt.tight_layout()
plt.savefig("conf_int.svg")
```

```python
## 1. Create boolean columns for significance
subset_corrupted_all['is_true_sig'] = subset_corrupted_all[TRUE_PVAL_COL] < 0.05
subset_corrupted_all['is_calc_sig'] = subset_corrupted_all[PVAL_COL] < 0.05

# 2. Dynamically find the True Moran's I ranges for ALL THREE groups
# --- Non-Significant ---
non_sig_mask = ~subset_corrupted_all['is_true_sig']
if non_sig_mask.any():
    min_i = subset_corrupted_all.loc[non_sig_mask, 'true_moran_i'].min()
    max_i = subset_corrupted_all.loc[non_sig_mask, 'true_moran_i'].max()
    label_nonsig = f"false significant"
else:
    label_nonsig = "false significant"

# --- Originally Significant (Positive) ---
sig_pos_mask = subset_corrupted_all['is_true_sig'] & (subset_corrupted_all['true_moran_i'] > 0)
if sig_pos_mask.any():
    min_pos = subset_corrupted_all.loc[sig_pos_mask, 'true_moran_i'].min()
    max_pos = subset_corrupted_all.loc[sig_pos_mask, 'true_moran_i'].max()
    label_pos = f"false non-significant (positive true Moran's I)"
else:
    label_pos = "false non-significant (positive true Moran's I)"

# --- Originally Significant (Negative) ---
sig_neg_mask = subset_corrupted_all['is_true_sig'] & (subset_corrupted_all['true_moran_i'] < 0)
if sig_neg_mask.any():
    min_neg = subset_corrupted_all.loc[sig_neg_mask, 'true_moran_i'].min()
    max_neg = subset_corrupted_all.loc[sig_neg_mask, 'true_moran_i'].max()
    label_neg = f"false non-significant (negative true Moran's I)"
else:
    label_neg = "false non-significant (negative true Moran's I)"


# 3. Create a custom palette mapping to these exact dynamic labels
custom_palette_3 = {
    label_neg: '#08519c',     # Dark Blue (Cold spots losing sig)
    label_nonsig: '#999999',  # Neutral Gray (Random noise gaining sig)
    label_pos: '#a50f15'      # Dark Red (Hot spots losing sig)
}

# 4. Calculate the 3 lines
results = []
for g in subset_corrupted_all['g_run'].unique():
    g_df = subset_corrupted_all[subset_corrupted_all['g_run'] == g]
    if len(g_df) == 0: continue
    
    mean_missing = g_df['real_missing'].mean() * 100
    
    # Line 1: Originally Non-Sig -> Measure % that GAINED significance
    g1 = g_df[~g_df['is_true_sig']]
    if len(g1) > 0:
        pct_faked = (g1['is_calc_sig'] == True).sum() / len(g1) * 100
        results.append({'mean_missing': mean_missing, 'Category': label_nonsig, 'Error Rate': pct_faked})
    
    # Line 2: Originally Sig (Positive) -> Measure % that LOST significance
    g2 = g_df[g_df['is_true_sig'] & (g_df['true_moran_i'] > 0)]
    if len(g2) > 0:
        pct_lost_pos = (g2['is_calc_sig'] == False).sum() / len(g2) * 100
        results.append({'mean_missing': mean_missing, 'Category': label_pos, 'Error Rate': pct_lost_pos})
        
    # Line 3: Originally Sig (Negative) -> Measure % that LOST significance
    g3 = g_df[g_df['is_true_sig'] & (g_df['true_moran_i'] < 0)]
    if len(g3) > 0:
        pct_lost_neg = (g3['is_calc_sig'] == False).sum() / len(g3) * 100
        results.append({'mean_missing': mean_missing, 'Category': label_neg, 'Error Rate': pct_lost_neg})

# Convert to DataFrame
plot_df = pd.DataFrame(results)

# 5. --- PLOTTING ---
fig, ax = plt.subplots(figsize=(8,7))

sns.lineplot(
    data=plot_df,
    x='mean_missing',
    y='Error Rate',
    hue='Category',
    palette=custom_palette_3,
    style='Category',
    # Solid for Lost Sig (Blue/Red), Dashed for Faked Sig (Gray)
    dashes={label_neg: '', label_pos: '', label_nonsig: ''},
    markers={label_neg: 'o', label_pos: 'o', label_nonsig: 's'},
    linewidth=2.5,
    markersize=8,
    alpha=0.85,
    ax=ax
)

# Formatting
ax.set_xlabel("edges missing (%)", fontsize=20)
ax.set_ylabel("significance misclassification (%)", fontsize=20)
#ax.set_title("Significance Stability under Network Degradation", fontsize=15)

ax.set_ylim(bottom=0)
ax.grid(True, linestyle=':', alpha=0.6)
sns.despine(ax=ax)

# Clean up Legend
sns.move_legend(ax, frameon=False,loc="upper left",title=None,alignment = "left",
    bbox_to_anchor=(0.02, 1))

plt.tight_layout()
plt.savefig("p_values_perclines.svg")
```

```python
# CHANGE THESE to your actual p-value column names if needed
PVAL_COL = 'p_value' 
TRUE_PVAL_COL = 'true_p_value'

# 1. Set up a single figure
fig, ax = plt.subplots(figsize=(8,7))

# 2. Extract the data for the ENTIRE dataset
y_true = subset_corrupted_all['true_moran_i']
y_calc = subset_corrupted_all['moran_i']      

pval_calc = subset_corrupted_all[PVAL_COL]
pval_true = subset_corrupted_all[TRUE_PVAL_COL]

# 3. Create boolean masks for the 4 categories
mask_tp = (pval_true < 0.05) & (pval_calc < 0.05)   # 1. Maintained Signal (True Positive)
mask_fn = (pval_true < 0.05) & (pval_calc >= 0.05)  # 2. Lost Signal (False Negative)
mask_fp = (pval_true >= 0.05) & (pval_calc < 0.05)  # 3. Fake Signal (False Positive)
mask_tn = (pval_true >= 0.05) & (pval_calc >= 0.05) # 4. True Non-Sig (True Negative)

# --- 4. PLOT THE COMMON, "EXPECTED" DOTS FIRST (zorder=1) ---
# We make these slightly transparent (alpha=0.3) and smaller (s=5) to form the background
ax.scatter(y_true[mask_tn], y_calc[mask_tn], color='firebrick', alpha=0.3, s=1, zorder=1, 
           label='true non-significant')
ax.scatter(y_true[mask_tp], y_calc[mask_tp], color='steelblue', alpha=0.3, s=1, zorder=1, 
           label='true significant')

# --- 5. PLOT THE RARE, "ERROR" DOTS ON TOP (zorder=2) ---
# We make these fully opaque, slightly larger (s=15), and add a black edge so they jump out
ax.scatter(y_true[mask_fn], y_calc[mask_fn], color='tomato', alpha=0.3, s=1, zorder=2,label='false non-significant')
ax.scatter(y_true[mask_fp], y_calc[mask_fp], color='deepskyblue', alpha=0.3, s=1, zorder=2,  label='false significant')

# 6. Draw standard X/Y axes through zero
ax.axhline(0, color='gray', linewidth=1, alpha=0.5)
ax.axvline(0, color='gray', linewidth=1, alpha=0.5)

# 7. Draw the perfect calculation diagonal line (y = x)
ax.axline((0, 0), slope=1, color='black', linewidth=1.5, linestyle='--', alpha=0.7)

# 8. Labels and Formatting
ax.set_xlabel("true Moran's I", fontsize=20)
ax.set_ylabel("calculated Moran's I", fontsize=20)

# Despine for a cleaner look
sns.despine(ax=ax)

# 9. Automatic Legend
# 'markerscale=2' makes the dots bigger inside the legend box so they are easy to read
ax.legend(frameon=False, markerscale=3,loc="upper left",fontsize=12,alignment = "left",
    bbox_to_anchor=(0.02, 1))

plt.tight_layout()
plt.savefig("p_value_dots.svg")
```

```python
truth_small = truth[["dataset", "variable", "moran_i", "moran_p"]].rename(
    columns={"moran_i": "true_moran_i", "moran_p": "true_moran_p"}
)

corrupted_list = []
for dataset in real_datasets:
    corrupted = pd.read_parquet(f"results/autocorrelation/real/random/global/{dataset}.parquet")
    g_true = graph.read_parquet(f"graphs/real/{dataset}/g_true.parquet")

    # true edge count, halved since libpysal stores both i->j and j->i for undirected graphs
    true_n_edges = g_true.n_edges / 2

    # merge in true moran values
    corrupted = corrupted.merge(truth_small, on=["dataset", "variable"], how="left")

    # deviation (signed) and absolute deviation
    corrupted["moran_deviation"] = corrupted["moran_i"] - corrupted["true_moran_i"]
    corrupted["moran_deviation_abs"] = corrupted["moran_deviation"].abs()

    # percentage of missing edges relative to true graph
    corrupted["true_n_edges"] = true_n_edges
    corrupted["pct_missing_edges"] = corrupted["missing_edges"] / true_n_edges * 100

    corrupted_list.append(corrupted)

all_corrupted = pd.concat(corrupted_list, ignore_index=True)
all_corrupted
```

```python
# true edge count for elections (undirected, so halve libpysal's count)
g_true_elections = graph.read_parquet("graphs/real/elections/g_true.parquet")
true_n_edges_elections = g_true_elections.n_edges / 2

truth_small = truth[["dataset", "variable", "moran_i", "moran_p"]].rename(
    columns={"moran_i": "true_moran_i", "moran_p": "true_moran_p"}
)

def add_truth_and_deviation(df, true_n_edges):
    df = df.merge(truth_small, on=["dataset", "variable"], how="left")
    df["moran_deviation"] = df["moran_i"] - df["true_moran_i"]
    df["moran_deviation_abs"] = df["moran_deviation"].abs()
    df["true_n_edges"] = true_n_edges
    df["pct_missing_edges"] = df["missing_edges"] / true_n_edges * 100
    return df

local_elections = add_truth_and_deviation(local_elections, true_n_edges_elections)
border_elections = add_truth_and_deviation(border_elections, true_n_edges_elections)

# unify region (local) and corruption_run (borders) into a single "group" column
local_elections = local_elections.rename(columns={"region": "group"})
border_elections = border_elections.rename(columns={"corruption_run": "group"})

elections_all = pd.concat([local_elections, border_elections], ignore_index=True)
elections_all
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator
from matplotlib.lines import Line2D
from matplotlib.legend_handler import HandlerTuple
import numpy as np

DEVIATION_COL = "moran_deviation"   # or "moran_deviation_abs"
fig, axes = plt.subplots(1, 2, figsize=(11, 5), sharey=True)

# --- borders: one line per run ---
ax = axes[0]
sub = guerry_extra[guerry_extra["corruption_method"] == "borders"]
for run, run_df in sub.groupby("run"):
    run_df = run_df.sort_values("pct_missing_edges")
    x = np.concatenate([[0], run_df["pct_missing_edges"]])
    y = np.concatenate([[0], run_df[DEVIATION_COL]])
    ax.plot(x, y, label=f"run {run}")
ax.axhline(0, color="gray", linewidth=0.6, linestyle="--")
ax.set_title("borders")
ax.set_xlabel("% edges missing")
ax.set_ylabel("Moran's I deviation")

# --- local: colour by region (high-contrast palette), alpha varies per run ---
ax = axes[1]
sub = guerry_extra[guerry_extra["corruption_method"] == "local"]

regions = sorted(sub["region"].dropna().unique())

# hand-picked high-contrast palette (Kelly/Glasbey-style), extend if you have more regions
distinct_colors = [
    "#e41a1c", "#377eb8", "#4daf4a", "#984ea3", "#ff7f00",
    "#a65628", "#f781bf", "#999999", "#1b9e77", "#d95f02",
    "#7570b3", "#66a61e", "#e6ab02", "#000000",
]
region_colors = {region: distinct_colors[i % len(distinct_colors)] for i, region in enumerate(regions)}

ALPHA_LOW, ALPHA_HIGH = 0.4, 1.0

for region, region_df in sub.groupby("region"):
    runs = sorted(region_df["run"].unique())
    alphas = np.linspace(ALPHA_LOW, ALPHA_HIGH, len(runs)) if len(runs) > 1 else [ALPHA_HIGH]
    for run, alpha in zip(runs, alphas):
        grp = region_df[region_df["run"] == run].sort_values("pct_missing_edges")
        x = np.concatenate([[0], grp["pct_missing_edges"]])
        y = np.concatenate([[0], grp[DEVIATION_COL]])
        ax.plot(x, y, color=region_colors[region], alpha=alpha, linewidth=1.3)

ax.axhline(0, color="gray", linewidth=0.6, linestyle="--")
ax.set_title("local")
ax.set_xlabel("% edges missing")

# legend: one entry per region, showing both a full-opacity and transparent swatch side by side
paired_handles, labels = [], []
for r in regions:
    full_line = Line2D([0], [0], color=region_colors[r], alpha=ALPHA_HIGH, lw=2)
    transparent_line = Line2D([0], [0], color=region_colors[r], alpha=ALPHA_LOW, lw=2)
    paired_handles.append((full_line, transparent_line))
    labels.append(r)

ax.legend(
    paired_handles, labels,
    handler_map={tuple: HandlerTuple(ndivide=None)},
    loc="upper left", fontsize=8, title="region"
)

# --- finer grid on both axes ---
for ax in axes:
    ax.xaxis.set_minor_locator(AutoMinorLocator())
    ax.yaxis.set_minor_locator(AutoMinorLocator())
    ax.grid(which="major", linewidth=0.6, alpha=0.6)
    ax.grid(which="minor", linewidth=0.3, alpha=0.3)

sns.despine()
fig.tight_layout()
plt.savefig("figs/guerry_moran_deviation_detail.png", dpi=600, bbox_inches="tight")
```
